```python
def binary_search(arr, search_item):
    N = len(arr)
    L = 0
    R = N-1 

    while L <= R:
        mid = (L+R)//2
        if arr[L] == search_item:
            print("Item found at index:", mid)
            return True
        elif arr[R] == search_item:
            print("Item found at index:", mid)
            return True

        elif arr[mid] == search_item:
            print("Item found at index:", mid)
            return True
        elif arr[mid] < search_item:
            L = mid + 1
        else:
            R = mid - 1
        
    return False
    
```

```python
arr = [-5,-3, 1, 3, 8, 9]
s = 1
print(binary_search(arr, s))
```
